11.	Обчисліть стандартне відхилення показників для кожної країни. Яка країна має найвищу варіативність у показниках між роками?

```python
df["Standard deviation"] = df[
    ["IMF (2026)", "World Bank (2025)", "United Nations (2024)"]
].std(axis=1)

df[["Country", "Standard deviation"]] \
    .sort_values("Standard deviation", ascending=False)
```

```python
max_std = df.loc[
    df["Standard deviation"].idxmax()
]

print("Країна з найвищою варіативністю:")
print(max_std[["Country", "Standard deviation"]])
```

12.	Визначення країни з найвищим та найнижчим показниками: Знайдіть країну з найвищим та найнижчим показниками у кожному з років (IMF (2026), World Bank (2025), United Nations (2024)).

```python
years = [
    "IMF (2026)",
    "World Bank (2025)",
    "United Nations (2024)"
]

for year in years:
    max_country = df.loc[df[year].idxmax(), "Country"]
    min_country = df.loc[df[year].idxmin(), "Country"]

    max_value = df[year].max()
    min_value = df[year].min()

    print(f"\n{year}")
    print(f"Найвищий: {max_country} — {max_value:,.0f} млн $")
    print(f"Найнижчий: {min_country} — {min_value:,.0f} млн $")
```

14.	Розрахуйте частку кожної країни в загальному значенні для кожного року (IMF (2026), World Bank (2025), United Nations (2024)).
 Як змінюються частки країн з часом (дати відповідь)?

```python
for year in years:
    total = df[year].sum()
    df[f"{year} share (%)"] = df[year] / total * 100

df[
    [
        "Country",
        "IMF (2026) share (%)",
        "World Bank (2025) share (%)",
        "United Nations (2024) share (%)"
    ]
].head(20)
```

15.	Візуалізуйте зміни в показниках для кожної країни за три роки на графіку. Які країни показують стабільне зростання або спад (дати відповідь)?

```python
top_countries = df.nlargest(15, "IMF (2026)")

plt.figure(figsize=(14, 8))

for _, row in top_countries.iterrows():
    plt.plot(
        years,
        row[years],
        marker="o",
        label=row["Country"]
    )

plt.xlabel("Year / Source")
plt.ylabel("GDP, million US$")
plt.title("Зміна показників ВВП за три роки")
plt.xticks(rotation=20)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")

plt.grid(True)
plt.show()
```

```python
for _, row in df.iterrows():
    values = row[years].values

    if values[0] > values[1] > values[2]:
        trend = "стабільний спад"
    elif values[0] < values[1] < values[2]:
        trend = "стабільне зростання"
    else:
        trend = "змішана динаміка"

    print(f"{row['Country']}: {trend}")
```
